```python
# ==========================================
# 0. GLOBALE KONSTANTEN & KONFIGURATION
# ==========================================
SYMBOL = "SILVER"
TIMEFRAME = "M30"
LIMIT = 3000
TZ_OFFSET_HOURS = 2

# ==========================================
# 1. SETUP & IMPORTE
# ==========================================
import sys
from pathlib import Path

PROJ_ROOT = Path(r"/")
if str(PROJ_ROOT) not in sys.path:
    sys.path.append(str(PROJ_ROOT))

%load_ext autoreload
%autoreload 2

from algos.chart_engine import load_candles, show_chart
from algos.ma_indicator import MAIndicator
from algos.grid_indicator import GridIndicator

# ==========================================
# 2. DATEN LADEN
# ==========================================
df_base = load_candles(
    symbol=SYMBOL,
    timeframe=TIMEFRAME,
    limit=LIMIT,
    tz_offset_hours=TZ_OFFSET_HOURS,
)
print(f"Daten geladen: {len(df_base)} Kerzen für {SYMBOL} ({TIMEFRAME}, Limit={LIMIT}).")
```

```python
# ==========================================
# 1. HIER INDIKATOREN & SIGNALE DEFINIEREN
# ==========================================
df_calc = df_base.copy()

# 0) MA-Instanz erstellen & berechnen
ma = MAIndicator(
    ma_type="TEMA",
    period=17,
    smoothing=5,
    alpha_factor=3.0,
)
df_calc = ma.apply(df_calc, generate_signals=True)

# 1) GridIndicator instanziieren
grid = GridIndicator({
    "step_size": 0.50,
    "steps_around": 4,
    "custom_levels": [64.30, 63.70],
    "visit_pct": 0.05,
    "grid_line_width": 1,
    "custom_line_width": 1,
    "line_color": "rgba(41, 121, 255, 0.85)",
    "use_time_filter": True,
    "time_window_mins": 5,
    "show_lines": True,
    "show_circles": True,
})

# ==========================================
# 2. AUTOMATISCHER PLOT
# ==========================================
show_chart(
    df_calc,
    symbol=SYMBOL,
    timeframe=TIMEFRAME,
    ma_indicator=ma,
    grid_indicator=grid,
    show_day_separators=True,
    visible_bars=350,
)
```

```python
# ==========================================
# DEBUG: TEXT- / UNICODE-KREIS TEST
# ==========================================
from lightweight_charts import JupyterChart
import pandas as pd

debug_chart = JupyterChart(width=1000, height=450)
debug_chart.layout(background_color="#131722", text_color="#d1d4dc")

# 1. Kerzen setzen (letzte 200 Bars)
sub_df = df_base.iloc[-200:].copy()
debug_chart.set(sub_df[["time", "open", "high", "low", "close"]])

# 2. Grid Indikator berechnen
grid_dbg = GridIndicator({
    "step_size": 0.50,
    "steps_around": 4,
    "custom_levels": [64.30, 63.70],
    "visit_pct": 0.05,
    "show_lines": True,
    "show_circles": True,
})
res_dbg = grid_dbg.calculate(sub_df)

# 3. Grid-Linien einzeichnen
t_start = sub_df["time"].iloc[0]
t_end = sub_df["time"].iloc[-1]
for gl in res_dbg.get("lines", []):
    l = debug_chart.create_line(color=gl["color"], width=gl["width"], price_line=False, price_label=False)
    l.set(pd.DataFrame({"time": [t_start, t_end], "value": [gl["price"], gl["price"]]}))

# 4. TEST: Text- / Punkt-Methoden prüfen
hits = res_dbg.get("hit_circles", [])
print(f"Gefundene Hits im Testbereich: {len(hits)}")

methods_tested = []

# METHODE 1: chart.text()
if hasattr(debug_chart, "text"):
    methods_tested.append("chart.text()")
    for h in hits[:30]:  # Erste 30 Hits
        try:
            debug_chart.text(
                time=h["time"],
                price=h["price"],
                text="●",
                color=h["color"]
            )
        except Exception as e:
            print(f"Fehler bei chart.text(): {e}")

# METHODE 2: Unicode Circle als Marker-Text auf der jeweiligen Gridline
# (Jede Gridline erhält ihre eigenen Treffer als Text-Zeichen)
hits_by_price = {}
for h in hits:
    p = round(float(h["price"]), 6)
    hits_by_price.setdefault(p, []).append(h)

for gl in res_dbg.get("lines", []):
    p_lvl = round(float(gl["price"]), 6)
    if p_lvl in hits_by_price:
        # Separate Trägerserie mit exakten Treffer-Punkten
        pts = pd.DataFrame(hits_by_price[p_lvl])
        pt_line = debug_chart.create_line(
            color=gl["color"],
            width=0,
            price_line=False,
            price_label=False
        )
        # Werte an den exakten Hit-Zeiten auf das Grid-Level setzen
        pt_data = pd.DataFrame({"time": pts["time"], "value": p_lvl})
        pt_line.set(pt_data)

        # Unicode-Symbol '●' als Text anheften
        for row in pts.itertuples():
            pt_line.marker(
                time=row.time,
                position="inside",
                shape="circle",
                color=row.color,
                text="●",
            )
        methods_tested.append(f"pt_line.marker(text='●') für Level {p_lvl}")

print(f"Getestete Rendering-Methoden: {methods_tested}")
debug_chart.load()
```
